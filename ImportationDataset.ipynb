```python
!pip install -q gdown

# Télécharger le Dataset de Google Drive

import gdown
import zipfile
import os

file_id = "1PIRDtyfUM3prHQKm_tdfw_-5bGeVDJTc"
url = f"https://drive.google.com/uc?id={file_id}"

output = "dataset.zip"

print("Téléchargement du dataset...")
gdown.download(url, output, quiet=False)

# Unzip Dataset

print("Extraction du dataset...")

with zipfile.ZipFile(output, 'r') as zip_ref:
    zip_ref.extractall("/content/")

# Detection automatique des dossiers

print("Contenu de /content :", os.listdir("/content"))

possible_dirs = [d for d in os.listdir("/content") if "brain" in d.lower()]

if len(possible_dirs) > 0:
    DATA_PATH = f"/content/{possible_dirs[0]}"
else:
    raise Exception("Dataset folder not found")

# Paths

train_dir = os.path.join(DATA_PATH, "Training")
test_dir  = os.path.join(DATA_PATH, "Testing")

# 6. Vérification

print("\nDATA PATH :", DATA_PATH)
print("Train classes :", os.listdir(train_dir))
print("Test classes  :", os.listdir(test_dir))
```
